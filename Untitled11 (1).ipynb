```python
import requests, re, json, time, random
from bs4 import BeautifulSoup
from dataclasses import dataclass, field
from typing import List, Dict, Any, Optional
from datetime import datetime
from ddgs import DDGS

# ---------- TOOL 1: Web Search (uses the ddgs package — reliable, no API key needed) ----------
def web_search_tool(query: str, max_results: int = 5, _simulate_failure: bool = False) -> Dict[str, Any]:
    """Searches the web and returns titles/snippets/links."""
    if _simulate_failure:
        # Deliberately induced failure for the assignment's robustness requirement
        raise TimeoutError(f"Simulated network timeout while searching: '{query}'")

    try:
        with DDGS() as ddgs:
            raw_results = list(ddgs.text(query, max_results=max_results))
    except Exception as e:
        # Wrap whatever ddgs/network throws into a consistent error type
        raise ConnectionError(f"Search backend failed for '{query}': {e}")

    results = [
        {"title": r.get("title", ""), "url": r.get("href", ""), "snippet": r.get("body", "")}
        for r in raw_results
    ]

    if not results:
        raise ValueError(f"No results found for query: '{query}'")

    return {"query": query, "results": results}


# ---------- TOOL 2: Calculator (safe arithmetic, used for scoring/aggregation) ----------
import ast, operator as op

_OPS = {ast.Add: op.add, ast.Sub: op.sub, ast.Mult: op.mul, ast.Div: op.truediv,
        ast.Pow: op.pow, ast.USub: op.neg}

def _eval_node(node):
    if isinstance(node, ast.Constant):
        return node.value
    if isinstance(node, ast.BinOp):
        return _OPS[type(node.op)](_eval_node(node.left), _eval_node(node.right))
    if isinstance(node, ast.UnaryOp):
        return _OPS[type(node.op)](_eval_node(node.operand))
    raise ValueError("Unsupported expression")

def calculator_tool(expression: str) -> Dict[str, Any]:
    """Safely evaluates an arithmetic expression."""
    tree = ast.parse(expression, mode="eval")
    result = _eval_node(tree.body)
    return {"expression": expression, "result": result}


TOOLS = {"web_search": web_search_tool, "calculator": calculator_tool}
```

```python
@dataclass
class Step:
    id: int
    description: str
    tool: str
    args: Dict[str, Any]
    status: str = "pending"      # pending | success | failed | recovered
    attempts: int = 0
    output: Any = None
    error: Optional[str] = None

@dataclass
class Trace:
    goal: str
    plan: List[Step] = field(default_factory=list)
    log: List[str] = field(default_factory=list)

    def note(self, msg):
        ts = datetime.now().strftime("%H:%M:%S")
        line = f"[{ts}] {msg}"
        self.log.append(line)
        print(line)


class PlanningModule:
    """Decomposes a high-level goal into concrete tool-using steps.
    Rule-based by default (no API key needed). Swap in an LLM call here
    (Anthropic/OpenAI) for more dynamic planning if desired."""

    def plan(self, goal: str, company: str) -> List[Step]:
        steps = [
            Step(1, f"Search for '{company}' company overview", "web_search",
                 {"query": f"{company} company overview what they do"}),
            Step(2, f"Search for '{company}' main competitors", "web_search",
                 {"query": f"{company} main competitors alternatives"}),
            Step(3, f"Search for recent '{company}' news", "web_search",
                 {"query": f"{company} latest news 2026", "_simulate_failure": True}),
            Step(4, "Compute a confidence/coverage score from sources gathered",
                 "calculator", {"expression": "(3 * 10 + 2 * 5) / 2"}),
        ]
        return steps


class ExecutionModule:
    """Executes steps, retries on failure, and self-corrects the approach."""

    def __init__(self, tools: Dict[str, Any], trace: Trace, max_retries: int = 2):
        self.tools = tools
        self.trace = trace
        self.max_retries = max_retries

    def run_step(self, step: Step):
        self.trace.note(f"STEP {step.id}: {step.description}")
        tool_fn = self.tools[step.tool]
        args = dict(step.args)

        for attempt in range(1, self.max_retries + 2):
            step.attempts = attempt
            try:
                self.trace.note(f"  -> calling tool '{step.tool}' (attempt {attempt}) args={ {k:v for k,v in args.items() if not k.startswith('_')} }")
                # Only actually simulate failure on the first attempt
                if attempt > 1:
                    args["_simulate_failure"] = False
                output = tool_fn(**args)
                step.output = output
                step.status = "success" if attempt == 1 else "recovered"
                self.trace.note(f"  -> OK ({step.status})")
                return
            except Exception as e:
                step.error = str(e)
                self.trace.note(f"  -> ERROR: {e}")
                if attempt <= self.max_retries:
                    backoff = 0.5 * attempt
                    self.trace.note(f"  -> self-correcting: retrying in {backoff}s "
                                     f"(reformulating query if applicable)")
                    if step.tool == "web_search" and "query" in args:
                        args["query"] = args["query"] + " 2026"  # reformulate
                    time.sleep(backoff)
                else:
                    step.status = "failed"
                    self.trace.note(f"  -> giving up after {attempt} attempts, marking step failed")
                    return

    def run_plan(self, steps: List[Step]):
        for step in steps:
            self.run_step(step)


class ReportingModule:
    """Synthesizes a structured final report from the execution trace."""

    def build(self, goal: str, company: str, steps: List[Step]) -> Dict[str, Any]:
        sources = []
        for s in steps:
            if s.tool == "web_search" and s.status in ("success", "recovered") and s.output:
                sources.extend(s.output.get("results", []))

        score = next((s.output["result"] for s in steps
                      if s.tool == "calculator" and s.output), None)

        failed_steps = [s.id for s in steps if s.status == "failed"]
        recovered_steps = [s.id for s in steps if s.status == "recovered"]

        summary_bullets = []
        for src in sources[:6]:
            if src.get("title"):
                summary_bullets.append(f"- **{src['title']}** — {src.get('snippet','')[:180]}")

        report = {
            "goal": goal,
            "company": company,
            "generated_at": datetime.now().isoformat(),
            "confidence_score": score,
            "steps_failed": failed_steps,
            "steps_recovered_via_retry": recovered_steps,
            "sources_collected": sources,
            "markdown_report": self._to_markdown(company, summary_bullets, score,
                                                   failed_steps, recovered_steps),
        }
        return report

    def _to_markdown(self, company, bullets, score, failed, recovered):
        md = [f"# Competitive Landscape Brief: {company}\n"]
        md.append("## Key Findings\n")
        md.extend(bullets if bullets else ["- No sources retrieved."])
        md.append(f"\n## Coverage Confidence Score: {score}\n")
        if recovered:
            md.append(f"*Note: step(s) {recovered} initially failed and were "
                       f"recovered automatically via retry/reformulation.*\n")
        if failed:
            md.append(f"*Warning: step(s) {failed} failed after all retries and "
                       f"were skipped — findings above are based on partial data.*\n")
        return "\n".join(md)
```

```python
class Agent:
    def __init__(self):
        self.planner = PlanningModule()
        self.reporter = ReportingModule()

    def run(self, goal: str, company: str):
        trace = Trace(goal=goal)
        trace.note(f"GOAL RECEIVED: {goal}")

        steps = self.planner.plan(goal, company)
        trace.plan = steps
        trace.note("PLAN:")
        for s in steps:
            trace.note(f"  {s.id}. {s.description}  [tool={s.tool}]")

        executor = ExecutionModule(TOOLS, trace)
        executor.run_plan(steps)

        trace.note("SYNTHESIZING FINAL REPORT...")
        report = self.reporter.build(goal, company, steps)
        trace.note("DONE.")
        return report, trace


# --- Run it ---
company = "Anthropic"   # change to any company name
goal = f"Given a company name, produce a short competitive-landscape brief using public web data. Company: {company}"

agent = Agent()
report, trace = agent.run(goal, company)

print("\n" + "="*60)
print(report["markdown_report"])
```
